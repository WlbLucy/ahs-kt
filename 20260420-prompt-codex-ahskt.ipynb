  核心原则

  - 如果你想证明“你的模型优于某个基线”，最好要在同一数据、同一预处理、同一划分、同一评测协议下复现。
  - 如果你只是想说明“已有研究大概做到什么水平”，可以引用论文原始结果，但必须明确写成文献报告结果，仅供参考，不构成严格可比对照。
  - 只要数据处理方式不同，结果就很可能不能直接比。这不是形式问题，而是任务定义已经变了。

  什么时候可以直接引用论文结果

  - 用在第二章相关工作，介绍某模型在原论文中的代表性表现。
  - 用在实验章前面的“文献参考表”，专门说明这些结果来自原论文或公开 benchmark。
  - 前提是你要标清楚：
      - 数据集版本
      - 预处理方式
      - 划分方式
      - 指标
      - 是否为原论文结果
  - 这种情况下，你的措辞应该是：
      - “文献报道结果”
      - “原论文在其设定下报告的结果”
      - “由于数据处理与评测协议不同，仅作参考，不作严格横向比较”

  什么时候不能直接拿来对比

  - 你想写“本方法优于 DKT/AKT/LBKT/DIMKT”
  - 你想在摘要、结论里下“超过现有主流方法”的结论

      - 多知识点展开方式不同
      - 是否做去泄露处理不同
      - 是否用了题目难度、提示、时间等额外特征不同

  第一性原理看，这件事为什么这么严格
  因为知识追踪的分数不是只由“模型强不强”决定，还强烈受以下因素影响：

  - 数据版本
  - 预处理
  - 评测协议
  - 信息泄露
  - 随机种子
  - 是否用额外特征

  也就是说，分数 = 模型能力 + 数据处理收益/损失 + 评测协议影响。
  如果你只拿别人论文里的分数来比，你比较的就不只是模型，还混进了别人的数据工程和评测设定。
  这会导致结论失真。
